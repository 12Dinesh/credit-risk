```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 5000

data = pd.DataFrame({
    "age": np.random.randint(21, 65, n),
    "income": np.random.normal(40000, 15000, n).astype(int),
    "utilization": np.random.uniform(0, 1, n),
    "cibil": np.random.randint(300, 900, n)
})

# create default logic (hidden real world truth)
prob_default = (
    0.4*(data["utilization"]) +
    0.3*(data["cibil"] < 650).astype(int) +
    0.2*(data["income"] < 30000).astype(int) +
    0.1*(data["age"] < 30).astype(int)
)

data["default"] = (np.random.rand(n) < prob_default).astype(int)
```

```python
data["age_bin"] = pd.cut(data["age"], bins=[20,30,40,50,65])
data["income_bin"] = pd.qcut(data["income"], 5)
data["util_bin"] = pd.qcut(data["utilization"], 5)
data["cibil_bin"] = pd.cut(data["cibil"], bins=[300,550,650,750,900])
```

```python
def woe_iv(df, feature, target):
    grouped = df.groupby(feature)[target].agg(["count","sum"])
    grouped.columns = ["total","bad"]
    grouped["good"] = grouped["total"] - grouped["bad"]

    dist_good = grouped["good"] / grouped["good"].sum()
    dist_bad  = grouped["bad"] / grouped["bad"].sum()

    woe = np.log(dist_good / dist_bad)
    iv = (dist_good - dist_bad) * woe

    result = grouped.copy()
    result["woe"] = woe
    result["iv"] = iv

    return result, iv.sum()
```

```python
for col in ["age_bin","income_bin","util_bin","cibil_bin"]:
    table, iv = woe_iv(data, col, "default")
    print("\n", col)
    print(table)
    print("IV =", iv)
```

```python
for col in ["age_bin","income_bin","util_bin","cibil_bin"]:
    woe_map = woe_iv(data, col, "default")[0]["woe"]
    data[col+"_woe"] = data[col].map(woe_map)
```
